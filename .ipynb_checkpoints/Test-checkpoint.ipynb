```python
import pandas as pd
```

```python
dicionario = pd.read_json('./predictions/few-shot/gpt-4/comic_styles_predictions.json').to_dict()
```

```python
for i in dicionario:
    item = dicionario[i]
    ht = item['humorous_text']
    md = item['model_name']
    an_cs = item['annotated_comic_styles']
    md_cs = item['model_comic_styles']
    
    if md == 'gpt-4' and len(ht.split()) <= 120 and an_cs['fun'] == 1 and md_cs['fun'] == 1:
        print(ht)
```
